Запуск обучения создаёт отдельный выпуск и обновляет models/current.json. БД не изменяется.


```python
import sys
from pathlib import Path

import pandas as pd

# Подключаем модули проекта при запуске из папки notebooks.
PROJECT_ROOT = Path.cwd().resolve()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

if not (PROJECT_ROOT / "config.ini").exists():
    raise RuntimeError("Запустите ноутбук из папки проекта или notebooks")

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.train import train_release
```

## Обучение и сохранение выпуска


```python
manifest = train_release()
```

## Сравнение validation-метрик и выбранная champion


```python
comparison = pd.DataFrame(
    [{"model": record["name"], **record["validation"]} for record in manifest["models"]]
).sort_values("f1", ascending=False)

display(comparison)
print("Рекомендованная champion:", manifest["champion_version"])
```
